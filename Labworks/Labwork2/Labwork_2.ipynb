```python
# HPC Labwork 2 : Get to know your GPU
# by Lucien Elyas Bedrane - 2640012

#!pip install numba

import numba
from numba import cuda

#Found on StackOverflow at this address : https://stackoverflow.com/questions/63823395/how-can-i-get-the-number-of-cuda-cores-in-my-gpu-using-python-and-numba/63833950#63833950
#Allows to find the nnumber of cores per SM depending on the compute capability of the current GPU used
cc_cores_per_SM_dict = {
    (2,0) : 32,
    (2,1) : 48,
    (3,0) : 192,
    (3,5) : 192,
    (3,7) : 192,
    (5,0) : 128,
    (5,2) : 128,
    (6,0) : 64,
    (6,1) : 128,
    (7,0) : 64,
    (7,5) : 64,
    (8,0) : 64,
    (8,6) : 128,
    (8,9) : 128,
    (9,0) : 128,
    (10,0) : 128,
    (12,0) : 128
    }

devices = numba.cuda.list_devices()

for e in devices:
    cc = e.compute_capability
    sms = getattr(e, 'MULTIPROCESSOR_COUNT')
    cores_per_sm = cc_cores_per_SM_dict.get(cc) # This could be replaced by 64 as there are only 64 cores per Streaming Multiprocessor on Google Collab T4 GPU
    total_cores = cores_per_sm * sms
    print("Device name: " + str(e.name)
      + ".\nCore info: Multiprocessor count is " + str(sms) + " and Core count is " + str(total_cores)
      + ".\nMemory Info: " + str(numba.cuda.current_context().get_memory_info()))
```
